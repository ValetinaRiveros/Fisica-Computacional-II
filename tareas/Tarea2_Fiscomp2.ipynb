```python
#Item 1
#(1)Considere una muestra de estrellas con las siguientes luminosidades de y distancias
Luminosidad = [1.0, 2.5, 0.8, 5.0, 10.0] #Lumiosidades, en unidades de luminosidad solar
distancias= [1.0, 2.5, 0.8, 5.0, 10.0] #Distancias, en parsec
```

```python
#(a) Convierta las listas de luminosidades y distancias en arreglos NumPy
Luminosidad_numpy = np.array(Luminosidad)
distancia_numpy = np.array(distancias)
#Se convierte la lista de luminosidad y distancias en arreglos con numpy.
```

```python
L_SI = Luminosidad_numpy *3.828*10**26 #convierto las luminisdades en Watts
D_SI = distancia_numpy*3.086*10**16 #Concierto las distancias en metros
```

```python
#(b)Calcule el flujo observado para cada estrella en Wm−2, defina una función para esto.
def flujo_observado(Luminosidad_numpy,distancia_numpy):
    F = Luminosidad_numpy/(4*np.pi*distancia_numpy**2)
    return F
#Se genera la funcion de flujo_observado para ocuparlo a lo largo del codigo.
```

```python
#(d)Determine el indice de la presenta el menor flujo observado
indice_min = flujo_observado(L_SI,D_SI)#Llamo a la funcion para ocuparla
np.argmin(indice_min)#Ocupo np.argmin para determinar el indice minimo de flujo
```

```python
#(e)Calcule el flujo promedio y la mediana del flujo de la muestra
f_obs = flujo_observado(L_SI,D_SI)#LLamo la funcion
np.mean(f_obs)#Ocupo np.mean para determinar el flujo promedio
```
